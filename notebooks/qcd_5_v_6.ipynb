```python
import re

import numpy as np
import uproot
import matplotlib.pyplot as plt
```

```python
f5 = uproot.open("../slimmed_root_5jets.root")
f6 = uproot.open("../slimmed_root_6jets.root")

# old = uproot.open("../root_files/cached/slimmed_root_6jets.root")
# old5 = uproot.open("../root_files/cached/slimmed_root_5jets.root")
```

```python
mix = uproot.open("../slimmed_mixed_QCD_2fps.root")
mix5 = uproot.open("../slimmed_mixed_QCD_5fps.root")
# low_mix = uproot.open("../stitched_hists_2.root")
```

```python
f5.keys()
```

```python
mix_norm = f6['njet'].to_hist().sum().value / mix['njet'].to_hist().sum().value
mix_norm
```

```python
mix5_norm = f6['njet'].to_hist().sum().value / mix5['njet'].to_hist().sum().value
mix5_norm
```

```python
f5['njet'].to_hist().plot1d(label='5 jets', histtype='step', color='red')
# old5['njet'].to_hist().plot1d(label='5 jets without 100-200 QCD slice', histtype='step', color='purple')
f6['njet'].to_hist().plot1d(label='6 jets', histtype='step', color='blue')
# old['njet'].to_hist().plot1d(label='6 jets without 100-200 QCD slice', histtype='step', color='orange')
(mix_norm * (mix['njet'].to_hist())).plot1d(label='Pseudo-Events (MC)', histtype='step', color='black')
(mix5_norm * (mix5['njet'].to_hist())).plot1d(label='Pseudo-Events (5 FPS MC)', histtype='step', color='green')

plt.legend()
# plt.savefig("plots_26_07_28/njet_w_100.pdf")
```

```python
f5['ht'].to_hist().plot1d(label='5 jets', histtype='step', color='red')
# old5['ht'].to_hist().plot1d(label='5 jets without 100-200 QCD slice', histtype='step', color='purple')
f6['ht'].to_hist().plot1d(label='6 jets ', histtype='step', color='blue')
# old['ht'].to_hist().plot1d(label='6 jets without 100-200 QCD slice', histtype='step', color='orange')
(mix_norm * (mix['ht'].to_hist())).plot1d(label='Pseudo-Events (MC)', histtype='step', color='black')
(mix5_norm * (mix5['ht'].to_hist())).plot1d(label='Pseudo-Events (5 FPS)', histtype='step', color='green')


plt.yscale('log')
plt.xlabel('HT [GeV]')
plt.legend()
# plt.savefig("plots_26_07_28/ht_w_100_6.pdf")
```

```python
norm_2 = f6['ht'].to_hist()[9:].sum().value / mix['ht'].to_hist()[9:].sum().value
norm_5 = f6['ht'].to_hist()[9:].sum().value / mix5['ht'].to_hist()[9:].sum().value

f5['ht'].to_hist()[9:].plot1d(label='5 jets', histtype='step', color='red')
# old5['ht'].to_hist()[9:].plot1d(label='5 jets without 100-200 QCD slice', histtype='step', color='purple')
f6['ht'].to_hist()[9:].plot1d(label='6 jets', histtype='step', color='blue')
# old['ht'].to_hist()[9:].plot1d(label='6 jets without 100-200 QCD slice', histtype='step', color='orange')
(norm_2 * (mix['ht'].to_hist()[9:])).plot1d(label='Pseudo-Events (MC)', histtype='step', color='black')
(norm_5 * (mix5['ht'].to_hist()[9:])).plot1d(label='Pseudo-Events (5 FPS)', histtype='step', color='green')
#(low_mix_norm * (low_mix['ht'].to_hist()[9:])).plot1d(label='Pseudo-Events (Low Stats)', histtype='step', color='green')

plt.yscale('log')
plt.xlabel('HT [GeV]')
plt.legend()

# plt.savefig("plots_26_07_28/ht_cut_with100.pdf")
```

```python
f5['jet_pt'].to_hist().plot1d(label='5 jets', histtype='step', color='red')
# old5['jet_pt'].to_hist().plot1d(label='5 jets without 100-200 QCD slice', histtype='step', color='purple')
f6['jet_pt'].to_hist().plot1d(label='6 jets', histtype='step', color='blue')
# old['jet_pt'].to_hist().plot1d(label='6 jets without 100-200 QCD slice', histtype='step', color='orange')

(mix_norm * (mix['jet_pt'].to_hist())).plot1d(label='Pseudo-Events (MC)', histtype='step', color='black')
(mix5_norm * (mix5['jet_pt'].to_hist())).plot1d(label='Pseudo-Events (5 FPS)', histtype='step', color='green')
#(low_mix_norm * (low_mix['jet_pt'].to_hist()))[].plot1d(label='Pseudo-Events (Low Stats)', histtype='step', color='green')

plt.yscale('log')
plt.xlabel('Jet p_{T} [GeV]')
plt.legend()

# plt.savefig("plots_26_07_28/lead_jet_pt_6.pdf")
```
